### Huggingface - Basic Inference

```python
from transformers import pipeline
```

```python
model = pipeline("sentiment-analysis")
model
```

```python
model("Today is an exciting day for learning AI")
```

```python
model("Today is a great day for learning AI... But, I choose DevOPS")
```

```python
model("Today is a great day for learning AI, but.. I choose DevOPS")
```

### Zero Shot Classification

```python
model = pipeline("zero-shot-classification")
```

```python
model("This is a course on Huggingface", candidate_labels=["politics", "sports", "education", "business"])
```

```python
model("This is a course on Huggingface", candidate_labels=["politics", "sports", "academics", "business"])
```

```python
model("This is a training module on Huggingface", candidate_labels=["politics", "sports", "academics", "business"])
```

### Using a specific model

```python
model = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-135M-Instruct")
```

```python
text = model("Write an essay. Topic: Gravity")
```

```python
print(text[0]['generated_text'])
```

### Wrong Use-case

```python
model('classify in to ["politics", "sports", "academics", "business"]: This is a course on Huggingface')
```

### Image Classification

```python
url = r"https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/pipeline-cat-chonk.jpeg"
```

```python
image_classifier = pipeline(
    task="image-classification",
    model="google/vit-base-patch16-224"
)
```
